```python
#rr no arrival
n=int(input("Enter the number of the process: "))
bt = list(map(int,input("Enter burst time: ").split()))
tq=int(input("Enter Time Quantum: "))
rt=bt[:]
wt=[0]*n
tat = [0]*n
wt[0]=0
time= 0
while True:
    finished= True
    for i in range(n):
        if rt[i]>0:
            finished= False
            if rt[i]>tq:
                time+=tq
                rt[i]-=tq
            else:
                time+= rt[i]
                wt[i]= time - bt[i]
                rt[i]=0
    if finished:
        break
for i in range(n):
    tat[i]= wt[i]+ bt[i]
print("\nProcess\tBT\tWT\tTAT")
for i in range(n):
    print(f"P{i+1}\t{bt[i]}\t{wt[i]}\t{tat[i]}")
```

```python
# SRTF Scheduling (Easiest Code)

n = int(input("Number of processes: "))
bt = list(map(int, input("Burst times: ").split()))
at = list(map(int, input("Arrival times: ").split()))

rt = bt[:]   # remaining times
wt = [0]*n
tat = [0]*n

time = 0
completed = 0

while completed < n:
    # find process with shortest remaining time among arrived
    shortest = -1
    for i in range(n):
        if at[i] <= time and rt[i] > 0:
            if shortest == -1 or rt[i] < rt[shortest]:
                shortest = i
    if shortest == -1:
        time += 1
        continue

    rt[shortest] -= 1
    time += 1

    if rt[shortest] == 0:
        completed += 1
        tat[shortest] = time - at[shortest]
        wt[shortest] = tat[shortest] - bt[shortest]

print("P\tBT\tAT\tWT\tTAT")
for i in range(n):
    print(f"P{i+1}\t{bt[i]}\t{at[i]}\t{wt[i]}\t{tat[i]}")

print("Average WT =", sum(wt)/n)
print("Average TAT =", sum(tat)/n)
```

```python
# Best Fit Memory Allocation

blocks = [100, 500, 200, 300, 600]
process = [212, 417, 112, 426]

allocation = [-1] * len(process)

for i in range(len(process)):
    best = -1

    for j in range(len(blocks)):
        if blocks[j] >= process[i]:
            if best == -1 or blocks[j] < blocks[best]:
                best = j

    if best != -1:
        allocation[i] = best
        blocks[best] -= process[i]

print("Process\tBlock")

for i in range(len(process)):
    if allocation[i] != -1:
        print(process[i], "\t", allocation[i] + 1)
    else:
        print(process[i], "\tNot Allocated")
```

```python
# Worst Fit Memory Allocation

blocks = [100, 500, 200, 300, 600]
process = [212, 417, 112, 426]

allocation = [-1] * len(process)

for i in range(len(process)):
    worst = -1

    for j in range(len(blocks)):
        if blocks[j] >= process[i]:
            if worst == -1 or blocks[j] > blocks[worst]:
                worst = j

    if worst != -1:
        allocation[i] = worst
        blocks[worst] -= process[i]

print("Process\tBlock")

for i in range(len(process)):
    if allocation[i] != -1:
        print(process[i], "\t", allocation[i] + 1)
    else:
        print(process[i], "\tNot Allocated")
```

```python
# Next Fit Memory Allocation

blocks = [100, 500, 200, 300, 600]
process = [212, 417, 112, 426]

allocation = [-1] * len(process)

j = 0

for i in range(len(process)):
    count = 0

    while count < len(blocks):

        if blocks[j] >= process[i]:
            allocation[i] = j
            blocks[j] -= process[i]
            break

        j = (j + 1) % len(blocks)
        count += 1

print("Process\tBlock")

for i in range(len(process)):
    if allocation[i] != -1:
        print(process[i], "\t", allocation[i] + 1)
    else:
        print(process[i], "\tNot Allocated")
```

```python
# Producer Consumer Problem using Queue

from queue import Queue
import threading
import time

q = Queue(5)

def producer():
    for i in range(5):
        q.put(i)
        print("Produced:", i)
        time.sleep(1)

def consumer():
    for i in range(5):
        item = q.get()
        print("Consumed:", item)
        time.sleep(1)

t1 = threading.Thread(target=producer)
t2 = threading.Thread(target=consumer)

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
# Resource Request Algorithm / bankers algorithm

available = [3, 3, 2]

maximum = [
    [7, 5, 3],
    [3, 2, 2],
    [9, 0, 2]
]

allocation = [
    [0, 1, 0],
    [2, 0, 0],
    [3, 0, 2]
]

request = [1, 0, 2]
p = 1   # Process number

need = []

for i in range(len(maximum)):
    row = []
    for j in range(len(available)):
        row.append(maximum[i][j] - allocation[i][j])
    need.append(row)

safe = True

for j in range(len(available)):

    if request[j] > need[p][j]:
        safe = False

    if request[j] > available[j]:
        safe = False

if safe:
    for j in range(len(available)):
        available[j] -= request[j]
        allocation[p][j] += request[j]
        need[p][j] -= request[j]

    print("Request Granted")
else:
    print("Request Denied")

print("Available:", available)
```

```python
# FIFO Page Replacement

pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frames = 3

memory = []
faults = 0

for page in pages:

    if page not in memory:

        if len(memory) < frames:
            memory.append(page)
        else:
            memory.pop(0)
            memory.append(page)

        faults += 1

    print(memory)

print("Total Page Faults =", faults)
```

```python
# LRU Page Replacement

pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frames = 3

memory = []
faults = 0

for page in pages:

    if page not in memory:

        if len(memory) < frames:
            memory.append(page)
        else:
            memory.pop(0)
            memory.append(page)

        faults += 1

    else:
        memory.remove(page)
        memory.append(page)

    print(memory)

print("Total Page Faults =", faults)
```

```python
# Dining Philosopher Problem

import threading
import time

n = 5
forks = [threading.Lock() for _ in range(n)]

def philosopher(i):
    left = i
    right = (i + 1) % n

    for _ in range(3):  # each philosopher eats 3 times

        print(f"Philosopher {i} is thinking")
        time.sleep(1)

        forks[left].acquire()
        forks[right].acquire()

        print(f"Philosopher {i} is eating")
        time.sleep(1)

        forks[left].release()
        forks[right].release()

t = []

for i in range(n):
    t.append(threading.Thread(target=philosopher, args=(i,)))
    t[i].start()

for i in t:
    i.join()

print("Done")
```
